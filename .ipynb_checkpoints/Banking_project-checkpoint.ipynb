## 1. Handle Missing Values

### •	Income → median by segment

```python
df['EmployerType'].value_counts()
```

```python
df.groupby('CustomerSegment')['MonthlyIncome'].median()
```

```python
df.groupby('EmployerType')['MonthlyIncome'].mean()
```

```python
df['MonthlyIncome']=df['MonthlyIncome'].fillna(df.groupby('EmployerType')['MonthlyIncome'].transform('median'))
```

```python
df['MonthlyIncome'].isnull().sum()
```

### •	Credit score → segment-based imputation

```python
df['CustomerSegment']=df['CustomerSegment'].fillna(df.groupby('CustomerSegment')['CreditScore'].transform('median'))
```

```python
df['CustomerSegment'].isna().sum()
```

```python
df['CustomerSatisfactionScore'].isna().astype(int)
```

```python
df['missing_customersatisfaction']=df['CustomerSatisfactionScore'].isna().astype(int) 
```

```python
df.groupby('CustomerSegment')['CustomerSatisfactionScore'].median()
```

```python
df['CustomerSatisfactionScore']=df['CustomerSatisfactionScore'].fillna(df.groupby('CustomerSegment')['CustomerSatisfactionScore'].transform('median'))
```

```python
df['CustomerSatisfactionScore'].isnull().sum()
```

#### 1. Identify High Profit Customers (Top 25%)
- We created a new dataset called high_profit by selecting the top 25% customers based on AccountBalance using quantile (0.75).

#### 2. Check Which Factors Dominate in High Profit Group 
#### ( Analyzed Key Characteristics)
1. CustomerSegment
2. NumberOfProducts
3. CreditScore
4. LoanFlag
5. InternetBanking
6. TenureYears

```python
high_profit['CustomerSegment'].value_counts().plot(kind='bar')
high_profit['CustomerSegment'].value_counts(normalize=True) * 100             #Slight dominance of SME customers
```

```python
high_profit['HighProfitProfile'] = (
    (high_profit['CustomerSegment'] == 'SME') &
    (high_profit['NumberOfProducts'] >= 3) &
    (high_profit['TenureYears'] > high_profit['TenureYears'].median()) &
    (high_profit['InternetBanking'] == 1)
)
```

```python
pd.crosstab(high_profit['CustomerSegment'],
            high_profit['HighProfitProfile']).plot(kind='bar', stacked=True)
plt.show()
```

```python
high_profit['NumberOfProducts'].value_counts(normalize=True) * 100              #Most profitable customers hold 3+ products
```

```python
high_profit['CreditScore'].describe()       

# Mean ≈ 574
# 75% customers below 709
# Profitability is not strongly driven by very high credit score
```

```python
df['LoanType'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title("Loan Type Distribution")
plt.ylabel('')
plt.show()

high_profit['LoanFlag'].value_counts(normalize=True) * 100            #Loans are almost equally distributed
```

```python
high_profit['InternetBanking'].value_counts(normalize=True) * 100          #Digital users slightly dominate profitable group
```

```python
high_profit['TenureYears'].describe()
```

```python
high_profit.loc['HighProfitProfile'] = (
    (high_profit['CustomerSegment'] == 'SME') &
    (high_profit['NumberOfProducts'] >= 3) &
    (high_profit['TenureYears'] > high_profit['TenureYears'].median()) &
    (high_profit['InternetBanking'] == 1)
)

high_profit['HighProfitProfile'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title("Profile Distribution Among High Profit Customers")
plt.ylabel('')
plt.show()
```

## 3. Tenure vs customer value?
1. #### Define Variables
    - TenureYears
2. #### Customer Value →
- AccountBalance
- NumberOfProducts
- MonthlyTransactionCount

```python
df[['TenureYears','AccountBalance']].corr()             # Check Basic Relationship (Tenure vs Balance)
# This is very weak positive correlation
```

```python
df.plot(
    kind='scatter',
    x='TenureYears',
    y='AccountBalance'
)
plt.title("Tenure vs Account Balance")           #There is no strong relationship between tenure and account balance.
plt.show()
```

#### Group Tenure into Categories

```python
df['TenureGroup'] = pd.cut(
    df['TenureYears'],
    bins=[0,5,10,15,20],
    labels=['0-5','5-10','10-15','15-20']
)
```

```python
df.groupby('TenureGroup',observed=True)['AccountBalance'].mean().plot(kind='bar')
plt.title("Average Balance by Tenure Group")                                 # Mean Balance by Tenure Group
plt.show()                                                # Balance increases slightly for long-tenure customers, but difference is small.
```

```python
df.groupby('TenureGroup',observed=True)['NumberOfProducts'].mean()                  #Newer customers (0–5 years) hold more products.
```

```python
df.groupby('TenureGroup',observed=True)['NumberOfProducts'].mean().plot(kind='bar')
plt.title("Products Held by Tenure Group")                         #New customers (0–5 years) hold slightly more products than older customers.
plt.show()
```

```python
df.groupby('TenureGroup',observed=True)['MonthlyTransactionCount'].mean()
```

1. Correlation between Tenure and Balance is very weak (0.049).
2. Scatter plot shows no clear trend — high balances exist at all tenure levels.
3. Long-tenure customers (15–20 years) have slightly higher average balance, but difference is small.
4. New customers (0–5 years) hold slightly more products.
5. Overall, tenure is not a strong driver of customer value.

```python
df.groupby('AccountDormantFlag')['ChurnFlag'].mean().plot(kind='bar', color=['skyblue', 'orange'])
plt.ylabel('Churn Rate')
plt.title('Churn Rate by Account Dormancy')
plt.show()
```

```python
pd.crosstab(df['AccountDormantFlag'], df['ChurnFlag']).plot(kind='bar', color=['skyblue', 'orange'])         
plt.show()
```

- Customers who stop using their accounts (dormant accounts) are more likely to leave the bank.
-  About 53% of dormant customers churned, compared to 45% of active customers.
-  This shows that inactivity increases the chances of churn.

```python
X = df[['InternetBanking',
        'MobileBanking',
        'LoginFrequencyMonthly',
        'MonthlyTransactionCount',
        'TenureYears',
        'NumberOfProducts',
        'CustomerSatisfactionScore']]

Y = df['ChurnFlag']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.3, random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, Y_train)
```

```python
coefficients = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': model.coef_[0]
})

print(coefficients)
```

- #### In this dataset, digital adoption does not reduce churn risk. 
- #### Customers using mobile banking and having higher activity levels show slightly higher churn rates.

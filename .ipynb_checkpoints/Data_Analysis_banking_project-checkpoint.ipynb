# Handling Missing Values

```python
# Income → median by segment
```

```python
df.MonthlyIncome.describe()
```

```python
df.MonthlyIncome.isna().sum()
```

```python
# df.groupby('BranchID')['CustomerSatisfactionScore'].median()
```

```python
df.groupby('CustomerSegment')['MonthlyIncome'].median()
```

```python
# df.groupby('EmployerType')['MonthlyIncome'].median()
```

```python
df['MonthlyIncome'] = df['MonthlyIncome'].fillna(df.groupby('CustomerSegment')['MonthlyIncome'].transform('median'))
```

```python
df.MonthlyIncome.isna().sum()
```

```python
# Credit score → segment-based imputation
```

```python
df['CreditScore'].isna().sum()
```

```python
df.groupby('RiskCategory')['CreditScore'].describe()
```

```python
# df.groupby('CustomerSegment')['CreditScore'].describe()
```

```python
df['CreditScore'] = df['CreditScore'].fillna(df.groupby('RiskCategory')['CreditScore'].transform('median'))
```

```python
df['CreditScore'].isna().sum()
```

```python
# Satisfaction → flag missingness bias
```

```python
df['CustomerSatisfactionScore'].isna().sum()
```

```python
df['SatisfactionMissing'] = df['CustomerSatisfactionScore'].isnull().astype(int)
```

```python
df['SatisfactionMissing']
```

```python
df
```

```python
df['CustomerSatisfactionScore'].isna().sum()
```

```python
df['CustomerSatisfactionScore'] = df['CustomerSatisfactionScore'].fillna(df['CustomerSatisfactionScore'].median())
```

```python
df['CustomerSatisfactionScore'].isna().sum()
```

3. Long-tenure customers (15–20 years) have slightly higher average balance, but difference is small.
4. New customers (0–5 years) hold slightly more products.
5. Overall, tenure is not a strong driver of customer value.

# 20.	Does relationship manager experience affect churn ?

- ### Does more RM experience → lower churn?

##### Average tenure for each Relationship Manager

```python
df.groupby('RelationshipManagerID')['ChurnFlag'].mean()
```

```python
# rm_churn = df.groupby('RelationshipManagerID')['ChurnFlag'].mean()

# rm_churn.plot(marker='o')
# plt.title("RM Experience vs Churn Rate")
# plt.ylabel("Churn Rate")
# plt.show()
```

```python
rm_analysis = df.groupby('RelationshipManagerID',observed=True).agg({
    'TenureYears':'mean',
    'ChurnFlag':'mean'
}).reset_index()
```

```python
conditions = [
    (rm_analysis['TenureYears'] >= 12) & (rm_analysis['ChurnFlag'] <= 0.30),
    (rm_analysis['TenureYears'] >= 8) & (rm_analysis['ChurnFlag'] <= 0.50)
]

choices = ['Good', 'Okay']

rm_analysis['ManagerPerformance'] = np.select(conditions, choices, default='Bad')
```

```python
rm_analysis['ManagerPerformance'].value_counts()
```

```python
rm_analysis['ManagerPerformance'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    title="Manager Performance Groups"
)
plt.ylabel('')
plt.show()
```

```python
rm_analysis.sort_values(by='TenureYears',ascending = False)
```
